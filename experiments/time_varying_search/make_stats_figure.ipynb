```python
import dill as pkl 
import matplotlib.pyplot as plt
import numpy as np
import glob

import seaborn as sns
```

```python
plt.rcParams["font.family"] = "Times New Roman"


fig, ax = plt.subplots(1,1, figsize=(4,2))
fnames = glob.glob('./data/time_opt_receding_horizon_trial*')
_times = []
_mean_vals = []
for fname in fnames:
    _log = pkl.load(open(fname, 'rb'))
    _times.append(_log.time[-1])
    _mean_vals.append(np.linalg.norm(_log.true_loc- _log.means[-1]))
print(np.mean(_mean_vals), np.std(_mean_vals), np.mean(_times))
ax.hist(_times, density=True, alpha=0.8, label=r'Time Opt.')

fnames = glob.glob('./data/time_opt_receding_horizon_0_01_trial*')
_times = []
_mean_vals = []
for fname in fnames:
    _log = pkl.load(open(fname, 'rb'))
    _times.append(_log.time[-1])
    _mean_vals.append(np.linalg.norm(_log.true_loc- _log.means[-1]))
print(np.mean(_mean_vals), np.std(_mean_vals), np.mean(_times))
ax.hist(_times, density=True, alpha=0.8, label=r'Time Opt.')


fnames = glob.glob('./data/receding_horizon_5_trial*')
_times = []
_mean_vals = []
for fname in fnames:
    _log = pkl.load(open(fname, 'rb'))
    _times.append(_log.time[-1])
    _mean_vals.append(np.linalg.norm(_log.true_loc- _log.means[-1]))
print(np.mean(_mean_vals), np.std(_mean_vals), np.mean(_times))
ax.hist(_times, density=True, alpha=0.8, label='Time Horiz.=5s')

fnames = glob.glob('./data/receding_horizon_10_trial*')
_times = []
_mean_vals = []
for fname in fnames:
    _log = pkl.load(open(fname, 'rb'))
    _times.append(_log.time[-1])
    _mean_vals.append(np.linalg.norm(_log.true_loc- _log.means[-1]))
print(np.mean(_mean_vals), np.std(_mean_vals), np.mean(_times))
ax.hist(_times, density=True, alpha=0.8, label='Time Horiz.=10s')



ax.spines[['right', 'top']].set_visible(False)

ax.set_ylabel('Prob. Density')
ax.set_xlabel('Time to Localize (s)')
ax.legend()

plt.savefig('time_analysis.pdf')
```
